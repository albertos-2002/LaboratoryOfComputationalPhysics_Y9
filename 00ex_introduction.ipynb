2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
#function definition
#not a permanent assignment with global

def makeSwapT(a,b):
    tmp = b
    b = a
    a = tmp
    return a,b

def makeSwapL(a,b):
    listT = [a,b]
    listT.reverse()
    return listT[0], listT[1]

def makeSwapN(a,b):
    return b,a
```

```python
a = 42
b = "cat"

print(makeSwapT(a,b))
print(makeSwapL(a,b))
print(makeSwapN(a,b))
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
#function definition

import math
def calcDist(a,b):
    #check of data type
    ax1 = a[0] - b[0]
    ax2 = a[1] - b[1]
    distance = math.sqrt( ax1**2 + ax2**2 )
    return distance
```

```python
u=(3,0)
v=(0,4)

print(calcDist(u,v))
```

11   12   13  14 \

```python
import math
endpoint = 100 #strictly less
triple_list = []
#2 for cicles to move on the upper part of the matrix
for f in range(1,endpoint):
    f_square = f**2
    for s in range(f,endpoint):
        c_square = f_square + s**2
        c_root = math.sqrt(c_square)
        #we need to find the integer
        AllZero = True 
        decimal_part = c_root % 1 #extract decimal part
        str_decimal = str(decimal_part) #string conversion
        str_decimal = str_decimal[2:] #remove 0.
        for char in str_decimal: #check if there is only zeros
            if char != "0" : AllZero = False 
        if AllZero and c_root < 100: #then append as a valid triple
            triple_list.append([f,s,int(c_root)])
    
print(triple_list)
print(len(triple_list))
```
