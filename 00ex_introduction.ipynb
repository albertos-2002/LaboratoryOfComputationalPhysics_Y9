2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
#function definition
#not a permanent assignment with global

def makeSwapT(a,b):
    tmp = b
    b = a
    a = tmp
    return a,b

def makeSwapL(a,b):
    listT = [a,b]
    listT.reverse()
    return listT[0], listT[1]
```

```python
a = 42
b = "cat"

print(makeSwapT(a,b))
print(makeSwapL(a,b))
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
#function definition

import math
def calcDist(a,b):
    #check of data type
    ax1 = a[0] - b[0]
    ax2 = a[1] - b[1]
    distance = math.sqrt( ax1**2 + ax2**2 )
    return distance
```

```python
u=(3,0)
v=(0,4)

print(calcDist(u,v))
```

```python
#we can use the same trick as before and use a dict to count values 
#the serch for the values 1 and increase a separate counter

ll = l
numdict = dict()
counter = 0
for n in ll:
    if n in numdict.keys():
        numdict[n] +=1
    else:
        numdict[n] = 1

for k in numdict.keys():
    if numdict[k] == 1:
        counter +=1

print("Unique numbers:   ", counter)
#for k in numdict.keys():
#    print(k, numdict[k])
```
