3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def addDecorator(ftd):
    def wrappedFunc(a):
        print("Hello")
        result = ftd(a)
        print(result)
    return wrappedFunc

def square(x):
    return x*x

decoratedSquare = addDecorator(square)
decoratedSquare(37)
````

````python
@addDecorator
def square(x):
    return x*x

square(37)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
#with no check for the integer type

def recursiveFactorial(n):
    if n == 0:
        return 1
    if n > 0: 
        return n*recursiveFactorial(n-1)
    
def factorial(n):
    result = 1
    if n != 0 :
        for i in range(1,n+1):
            result *= i
    return result
````

````python
print("Recursive -----------------------")
print(recursiveFactorial(0))
print(recursiveFactorial(4))
print("Not recursive -------------------")
print(factorial(0))
print(factorial(4))
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
````

````python
import math
#lambda function (result)
lambdaFunctionList = [
    lambda r: 2*math.pi*r,
    lambda r: math.pi * r**2,
    lambda r: (4/3)*math.pi * r**3
    ]
volumeCalculator = [
    [ lambdaFunctionList[0](r),lambdaFunctionList[1](r),lambdaFunctionList[2](r) ]
    for r in radii
]
print(volumeCalculator)

#lambda function (as function)
functionVolume = [
    [
        lambda r: 2*math.pi*r,
        lambda r: math.pi * r**2,
        lambda r: (4/3)*math.pi * r**3        
    ]
    for k in range(1)
]
print(functionVolume)
````

````python
mass = []
for r1 in radii:
    for index, d1,d2,d3 in zip(range(3), densities["Al"], densities["Fe"], densities["Pb"]):
        mass.append([
            functionVolume[0][index](r1)*d1,
            functionVolume[0][index](r1)*d2,
            functionVolume[0][index](r1)*d3
        ])
print("Mass of the Al            Fe                 Pb")
print("r1: ", mass[0],"Linear")
print("r1: ", mass[1],"Area")
print("r1: ", mass[2],"Volume")
print("r2: ", mass[3],"Linear")
print("r2: ", mass[4],"Area")
print("r2: ", mass[5],"Volume")
print("r3: ", mass[6],"Linear")
print("r3: ", mass[7],"Area")
print("r3: ", mass[8],"Volume")
````

````python
#addition to the class
class Dog:
    species = 'mammal' #shared by all - only one

    def __init__(self, name, age):
        self.name = name #belong to the class object
        self.age = age
        self.is_hungry = True

    def eat(self):
        self.is_hungry = False

    def isHungry(self):
        return self.is_hungry

    def description(self):
        return "{} is {} years old".format(self.name, self.age)
        
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)
````

````python
tomClass = Dog("Tom",6)
flatcherClass = Dog("Flatcher",7) 
larryClass = Dog("Larry",9)

classList = [tomClass, flatcherClass, larryClass]

print("Presentation: ------------")
for el in classList:
    print(el.description())

print("Hungry ??: --------------")
flagH = True
for el in classList:
    flagH = flagH and el.isHungry()
if flagH : print("My dogs are hungry.")

print("Let's eat: --------------")
for el in classList:
    el.eat()
#    print("Eating done")

print("Full ??: ----------------")
flagNH = True
for el in classList:
    flagNH = flagNH and el.isHungry()
if not flagNH : print("My dogs are not hungry.")
````
