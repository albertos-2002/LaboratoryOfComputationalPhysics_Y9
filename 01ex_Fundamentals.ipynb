2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
import copy
x = 5

def func(arg):
    int_arg = copy.copy(arg)
    for i in range(5):
        int_arg.append(i)
    return int_arg

alist = [1,2,3]
ans = func(alist)
print (ans)
print (alist) # alist has NOT changed!
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def addDecorator(ftd):
    def wrappedFunc(a):
        print("Hello")
        result = ftd(a)
        print(result)
    return wrappedFunc

def square(x):
    return x*x

decoratedSquare = addDecorator(square)
decoratedSquare(37)
````

````python
@addDecorator
def square(x):
    return x*x

square(37)
````
